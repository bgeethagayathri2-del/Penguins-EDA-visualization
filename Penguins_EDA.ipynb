```python
# Exploratory Data Analysis (EDA): Palmer Penguins
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

# 1. Load and inspect
df = sns.load_dataset("penguins")
print("Shape:", df.shape)
print(df.head())
print("\nMissing values:")
print(df.isnull().sum())
print("\nSummary statistics:")
print(df.describe().round(1))

# 2. Clean: remove rows with missing values
before = len(df)
df = df.dropna().reset_index(drop=True)
print(f"\nRows before cleaning: {before}, after: {len(df)}")

# 3. How many penguins of each species live on each island?
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="island", hue="species")
plt.title("Penguins by island and species")
plt.show()

# 4. Body mass by species (distribution)
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x="body_mass_g", hue="species", kde=True)
```
